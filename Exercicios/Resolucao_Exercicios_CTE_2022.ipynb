## 1. Checkpoint 1 (Problema 2): Preço Final da Gasolina Comum (Gasolina C)
**Enunciado:** Calcular o preço final da gasolina C na bomba com base nos custos da gasolina A na refinaria, etanol anidro, margem da distribuidora (5%), tributos federais CIDE/PIS/COFINS (20%), custos/lucros de revenda (15%) e ICMS estadual (25%).

```python
# Resolução Oficial do Problema 2 (1_Checkpoint CTFE v2.pdf)
def calcular_preco_gasolina(valor_gasolinaA, valor_etanol_anidro):
    valor_gasolinaC = valor_gasolinaA * 0.73 + valor_etanol_anidro * 0.27
    valor_distribuidora = valor_gasolinaC * 0.05
    valor_tributos_federais = (valor_gasolinaC + valor_distribuidora) * 0.20
    custos_lucros = (valor_gasolinaC + valor_distribuidora + valor_tributos_federais) * 0.15
    icms = (valor_gasolinaC + valor_distribuidora + valor_tributos_federais + custos_lucros) * 0.25
    valor_gasolina_final = valor_gasolinaC + valor_distribuidora + valor_tributos_federais + custos_lucros + icms
    
    return {
        'Gasolina C (Mistura)': valor_gasolinaC,
        'Distribuidora (5%)': valor_distribuidora,
        'Tributos Federais (20%)': valor_tributos_federais,
        'Custos e Lucros Revenda (15%)': custos_lucros,
        'ICMS Estadual (25%)': icms,
        'Preço Final no Posto': valor_gasolina_final
    }

# Exemplo de Teste do Enunciado Oficial:
res = calcular_preco_gasolina(valor_gasolinaA=3.86, valor_etanol_anidro=3.63)
for k, v in res.items():
    print(f'{k:32s}: R$ {v:.2f}')
```

## 2. Checkpoint 1 (Problema 3): Fatura de Energia Elétrica (TUSD, TE, PIS, COFINS e CIP)
**Enunciado:** Cálculo da fatura de energia a partir das leituras anterior e atual do medidor residencial.

```python
# Resolução Oficial do Problema 3 (questao3CTFE(1).py)
def calcular_fatura_energia(leitura_ant, leitura_atual):
    consumo = leitura_atual - leitura_ant
    tusd = consumo * 0.40942
    te = consumo * 0.38317
    pis_pasep = consumo * 0.01
    cofins = consumo * 0.0462
    cip = 24.01
    total_pagar = tusd + te + pis_pasep + cofins + cip
    
    return {
        'Consumo (kWh)': consumo,
        'TUSD (R$)': tusd,
        'TE (R$)': te,
        'PIS/PASEP (R$)': pis_pasep,
        'COFINS (R$)': cofins,
        'CIP (R$)': cip,
        'Total a Pagar (R$)': total_pagar
    }

res_energia = calcular_fatura_energia(57521, 57761)
for k, v in res_energia.items():
    print(f'{k:25s}: {v:.2f}')
```

## 3. Jogo Jokenpô (Pedra, Papel ou Tesoura)
**Enunciado:** Implementação interativa do jogo clássico contra o computador utilizando estruturas de repetição `while True`, condicionais encadeadas e o módulo `random.choice`.

```python
from random import choice

def avaliar_jokenpo(jogador, computador):
    if jogador == computador:
        return 'Empate'
    elif (jogador == 'PEDRA' and computador == 'TESOURA') or \
         (jogador == 'PAPEL' and computador == 'PEDRA') or \
         (jogador == 'TESOURA' and computador == 'PAPEL'):
        return 'Jogador Venceu'
    else:
        return 'Computador Venceu'

# Simulação de 5 rodadas automáticas de teste:
opcoes = ['PEDRA', 'PAPEL', 'TESOURA']
print('Simulação de Partidas de Jokenpô:')
for i in range(1, 6):
    jog = choice(opcoes)
    comp = choice(opcoes)
    res = avaliar_jokenpo(jog, comp)
    print(f'Rodada {i}: Jogador [{jog}] vs Computador [{comp}] -> {res}')
```

## 4. Checkpoint 2: Cálculo de Média Semestral e Anual com Exame
**Enunciado:** Sistema acadêmico de notas calculando médias parciais e situação final de aprovação.

```python
# Resolução do Checkpoint 2 (ckeck2_questao_1ECB.py)
def calcular_media_anual(cp1_1, cp2_1, ent1_1, ent2_1, gs_1, cp1_2, cp2_2, ent1_2, ent2_2, gs_2, nota_exame=None):
    med_trab_1 = (cp1_1 + cp2_1 + ent1_1 + ent2_1) / 4.0
    med_sem1 = med_trab_1 * 0.4 + gs_1 * 0.6
    
    med_trab_2 = (cp1_2 + cp2_2 + ent1_2 + ent2_2) / 4.0
    med_sem2 = med_trab_2 * 0.4 + gs_2 * 0.6
    
    media_parcial = med_sem1 * 0.4 + med_sem2 * 0.6
    
    if media_parcial >= 60.0:
        situacao = 'Aprovado Direto'
        media_final = media_parcial
    elif media_parcial >= 40.0 and nota_exame is not None:
        media_final = (media_parcial + nota_exame) / 2.0
        situacao = 'Aprovado com Exame' if media_final >= 60.0 else 'Reprovado no Exame'
    else:
        situacao = 'Reprovado'
        media_final = media_parcial
        
    return {
        'Média 1º Semestre': med_sem1,
        'Média 2º Semestre': med_sem2,
        'Média Parcial Anual': media_parcial,
        'Média Final': media_final,
        'Situação': situacao
    }

res_med = calcular_media_anual(60, 50, 50, 60, 60, 50, 70, 60, 60, 55, nota_exame=80)
for k, v in res_med.items():
    if isinstance(v, float):
        print(f'{k:25s}: {v:.1f}')
    else:
        print(f'{k:25s}: {v}')
```
